```python
creditdf=creditdf[['Car_Owner','Propert_Owner','Annual_income','EDUCATION', 'label']]
```

```python
creditdf
```

```python
creditdf['Car_Owner']=labelencoder.fit_transform(creditdf['Car_Owner'])
creditdf['Propert_Owner']=labelencoder.fit_transform(creditdf['Propert_Owner'])
creditdf['EDUCATION']=labelencoder.fit_transform(creditdf['EDUCATION'])
```

```python
creditdf
```

```python
X=creditdf.iloc[:,:-1]
```

```python
y=creditdf.iloc[:,-1]
```

```python
X_train, x_test,y_train,y_test=train_test_split(X,y, test_size=0.2)
```

```python
scaler=StandardScaler()
scaled_X_train=scaler.fit_transform(X_train)
scaled_x_test=scaler.fit_transform(x_test)

```

```python
def modelperformance (prediction):
    print("Accuracy score on model is {}".format (accuracy_score    (y_test, prediction)))
```

```python
logreg=LogisticRegression()
logreg.fit(scaled_X_train,y_train)
log_reg_pred=logreg.predict(scaled_x_test)

modelperformance(log_reg_pred)
```

```python
# kknnn neighbors

param_grid={"n_neighbors":[3,5,7,9,11],
           "weights":["uniform","distance"],
           "algorithm":["auto","kd_tree"]}
gridkn= GridSearchCV(KNeighborsClassifier(), param_grid, cv=2)
```

```python
gridkn.fit(scaled_X_train,y_train)
```

```python
gridkn.best_params_
```

```python
knpreds=gridkn.predict(scaled_x_test)
modelperformance(knpreds)
```
